```python
import tensorflow as tf
from tensorflow.keras import layers, Model

def build_autoencoder(input_shape):
    """
    Builds and returns the training and encoder models.
    """
    num_neurons = 16
    decoder_neurons = 16
    #mask = layers.Masking(mask_value=0.0)
    encoder_lstm = layers.LSTM(num_neurons,
                               return_sequences=True,
                               return_state=True,
                               name="encoder_lstm")
    
    decoder_lstm = layers.LSTM(decoder_neurons,
                               return_sequences=True,
                               return_state=True,
                               name="decoder_lstm")
    
    #attention_layer = layers.AdditiveAttention(name="attention")
    #concat_layer = layers.Concatenate(axis=-1, name="concat_attention")
    output_dense_layer = layers.Dense(input_shape[1], name="output_dense", activation="linear")
    output_time_dist = layers.TimeDistributed(output_dense_layer, name="output_time_distributed")


    inputs = tf.keras.Input(shape=(None,32), name="init_input")
    #x = mask(inputs)
    encoder_out, h, c = encoder_lstm(inputs)
    states = [h, c]

    target = inputs
    decoder_lstm_outputs, _, _ = decoder_lstm(target, initial_state=states)
    
    # attention_context = attention_layer([decoder_lstm_outputs, encoder_out])
    # decoder_combined_context = concat_layer([attention_context, decoder_lstm_outputs])
    decoder_pred_outputs = output_time_dist(decoder_lstm_outputs)

    training_model = Model(inputs=inputs, outputs=decoder_pred_outputs, name="Training_AutoEncoder")

    encoder_model = Model(inputs=inputs, outputs=[encoder_out] + states, name="ENCODER")
    
    return training_model, encoder_model

def train_model(training_model, X_train):
    """
    Compiles and fits the training model.
    """
    optim = tf.keras.optimizers.Adam(learning_rate=0.01)
    training_model.compile(optimizer=optim, loss="mse")
    training_model.summary()
    
    # For an autoencoder, the input (X_train) is also the target (y_train)
    training_model.fit(X_train, X_train,  
                       batch_size=32, 
                       epochs=15, 
                       verbose=True, 
                       validation_split=0.2)
    
    training_model.save_weights("C:/CodingWorkspace/Python/WunderRNN/LSTM_Pretrained.weights.h5")

# --- How to use it ---
# Assuming input_shape = (timesteps, features)
# For example: input_shape = (100, 32) 
# X_train = ... (your data)

# 1. Build the models
# training_model, encoder_model = build_autoencoder(input_shape)

# 2. Train the autoencoder
# train_model(training_model, X_train)

# 3. You can now use the encoder_model for inference
# context_vector = encoder_model.predict(some_new_data)
```

```python
trainer , encoder = build_autoencoder([1000,32])
train_model(trainer,seqs)
```
