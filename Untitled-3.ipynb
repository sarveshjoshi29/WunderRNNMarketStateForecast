```python
import tensorflow.keras.backend as K
def r2_keras(y_true, y_pred):
    """
    Custom R2-score metric for 3D tensors (batch, time_steps, features).
    Flattens the time steps dimension before calculating R2.
    """
    # Reshape tensors from (batch, time_steps, features) to (batch * time_steps, features)
    y_true_reshaped = K.reshape(y_true, (-1, K.shape(y_true)[-1]))
    y_pred_reshaped = K.reshape(y_pred, (-1, K.shape(y_pred)[-1]))

    # Calculate Sum of Squares Residual (SS_res)
    SS_res = K.sum(K.square(y_true_reshaped - y_pred_reshaped))

    # Calculate Total Sum of Squares (SS_tot)
    # Mean is calculated across all flattened samples
    SS_tot = K.sum(K.square(y_true_reshaped - K.mean(y_true_reshaped)))

    # Calculate R2 score, add epsilon to prevent division by zero
    r2 = 1 - SS_res / (SS_tot + K.epsilon())

    return r2
```

```python
class AutoEncoder(Model):
  def __init__(self,dims,input_shape,flatten = True):
    super(AutoEncoder,self).__init__()
    self.encode = tf.keras.Sequential(
        [
            layers.Flatten(),
            layers.Dense(dims,activation = 'sigmoid')
        ]
    )
    #print(int(np.prod(input_shape)))
    self.decode = tf.keras.Sequential(
        [

          layers.Dense(int(np.prod(input_shape)),activation = 'sigmoid'),
          layers.Reshape(input_shape)
        ]
        )

  def call(self,x):
    encoded = self.encode(x)
    decoded = self.decode(encoded)
    return decoded

class ConstructorAE():
  def __init__(self,dims,input_shape):
    self.dims = dims
    self.num_layers = len(dims)
    self.layers = []
    self.encoders = []
    for i in range(self.num_layers):
      if i == 0:
        build_shape = input_shape
        temp = AutoEncoder(dims[i],build_shape)
      else :
        build_shape = (self.dims[i-1],1)
        temp = AutoEncoder(dims[i],build_shape)
      self.layers.append(temp)

  def fit(self,x):

   # print(self.layers)
    data = x
    for i in range(len(self.layers)):
      print(f"preparing layer {i+1}")
      obj= self.layers[i]
      custom_opti = tf.keras.optimizers.Adam(learning_rate = 0.00001)
      obj.compile(optimizer= custom_opti,loss='mse')
      obj.fit(data,data,epochs = 13,verbose = 1)
      self.encoders.append(obj.encode)
      data = obj.encode(data).numpy()
    return tf.keras.Sequential(self.encoders)


  def transform(self,X_test):
    return self.encoder_layer(X_test)

```
