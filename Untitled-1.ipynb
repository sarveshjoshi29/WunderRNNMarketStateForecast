```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers
from tensorflow.keras.models import Model
from keras.layers import InputLayer,Input
import random
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.metrics import R2Score

SEED = 29

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)
```

```python
def create_model(input_shape,window_size = 20,latent_dims = 20):
    lstm_1 = layers.LSTM(64,return_sequences=True, return_state=False,activation='tanh')
    lstm_2 = layers.LSTM(latent_dims,return_state= False,activation='tanh')

    attention_layer = layers.AdditiveAttention(name="attention")
    concat_layer = layers.Concatenate(axis=-1, name="concat_attention")
    output_dense_layer = layers.Dense(input_shape[1], name="output_dense",activation="linear")

    inputs = Input(input_shape,name = "main_input")
    short_term_data = Input(input_shape,name = "short_term_input")
    x = lstm_1(inputs)
    x = lstm_2(x)

    qvec = layers.Reshape((1,latent_dims),name = "reshape_layer")(x)

    attention_context = attention_layer([qvec,short_term_data])
    attention_context_flat = layers.Reshape((-1,), name="flatten_attention")(attention_context)
    combined_context = concat_layer([attention_context_flat , x])
    out = output_dense_layer(combined_context)
    
    model = Model(inputs=[inputs, short_term_data],outputs=out)
    
    return model
    
def train_model(model,X_train_combined,y_train):

    optim = tf.keras.optimizers.Adam(learning_rate=0.01)
    model.compile(optimizer = optim,loss = "mse")
    model.fit(
    X_train_combined, 
    y_train, 
    epochs=5, 
    batch_size=32,
    validation_split=0.2,
    verbose = True
)



    


```

```python
import numpy as np

def create_sliding_windows(data, window_size=20, start_index=0):
    """
    Converts a 3D array of sequences into a sliding window
    format for training an LSTM model.
    
    This function creates (X, y) pairs where X is a window of
    'window_size' timesteps and y is the single timestep
    immediately following the window.

    Args:
        data (np.array): The input data of shape 
                         (num_samples, total_sequence_length, num_features).
        window_size (int): The number of timesteps to use for each
                           input window (e.g., 20).
        start_index (int): The starting timestep index to begin
                           creating windows. Defaults to 0.

    Returns:
        tuple: A tuple containing ([X_main, X_short], y)
               - X_main (np.array): Shape (new_samples, window_size, features)
               - X_short (np.array): A copy of X_main for the attention input.
               - y (np.array): Shape (new_samples, features)
    """
    print(f"Original data shape: {data.shape}")
    print(f"Using window size: {window_size}")
    print(f"Starting window creation from index: {start_index}")

    # Get dimensions
    num_seqs, total_len, num_features = data.shape
    
    # Calculate the number of samples we can get from a single sequence
    # The last possible starting index is (total_len - window_size - 1)
    end_point = total_len - window_size
    
    if start_index >= end_point:
        print(f"Error: start_index ({start_index}) is too large. "
              f"With total_len {total_len} and window_size {window_size}, "
              f"the last possible start is {end_point - 1}.")
        # Return empty arrays in the expected format
        empty_x_shape = (0, window_size, num_features)
        empty_y_shape = (0, num_features)
        empty_x = np.array([], dtype=data.dtype).reshape(empty_x_shape)
        empty_y = np.array([], dtype=data.dtype).reshape(empty_y_shape)
        return [empty_x, empty_x], empty_y
        
    samples_per_seq = total_len - window_size - start_index
    
    # Calculate the total number of samples that will be generated
    num_new_samples = num_seqs * samples_per_seq
    
    print(f"Generating {num_new_samples} training samples...")

    # Pre-allocate NumPy arrays for efficiency
    # This is much faster than appending to lists
    X_shape = (num_new_samples, window_size, num_features)
    y_shape = (num_new_samples, num_features)
    
    X_main_train = np.zeros(X_shape, dtype=data.dtype)
    y_train = np.zeros(y_shape, dtype=data.dtype)
    
    # Use a counter to fill the new arrays
    idx = 0
    for i in range(num_seqs):
        # Get one sequence, e.g., shape (1000, 32)
        sequence = data[i]
        
        # Slide the window across this sequence, starting from start_index
        for t in range(start_index, total_len - window_size):
            # The window is [t]...[t + window_size - 1]
            window = sequence[t : t + window_size]
            
            # The target is the step right after the window: [t + window_size]
            target = sequence[t + window_size]
            
            # Store in the pre-allocated arrays
            X_main_train[idx] = window
            y_train[idx] = target
            idx += 1
            
        if (i + 1) % 50 == 0 or i == num_seqs - 1:
            print(f"  Processed {i + 1} / {num_seqs} sequences...")

    # For your model, X_main and X_short are identical
    X_short_train = X_main_train
    
    print("Data generation complete.")
    
    # Return in the format model.fit() expects
    return [X_main_train, X_short_train], y_train
```

```python
X_train_combined , y_train = create_sliding_windows(seqs,20,99)
```

```python
model = create_model([None,32],20,32)
model.summary()
train_model(model,X_train_combined,y_train)
```
